# Chapter.7 Queue

### **1. 큐의 개념**
- 먼저 들어간 데이터가 먼저나오는 FIFO(First In First Out)구조
- 'push'를 통해 삽입, 'pop'을 통해 추출



### **2. 큐의 ADT**
- 연산
    - boolean isFull() : 큐에 들어있는 데이터 개수가 maxsize 인지를 확인
    - boolean isEmpty() : 큐에 들어있는 데이터가 없는지 확인 -> front == rear를 확인
    - void push(item) : 큐에 데이터 푸시
    - itemType pop() : 큐의 마지막 데이터 팝하고, 데이터 반환
- 상태
    - int front : 가장 마지막에 팝한 위치를 기록, 아무 데이터가 없는 초기값은 -1
    - int rear : 가장 최근에 푸시한 데이터의 위치를 기록, 아무 데이터가 없는 초기값은 -1
    - itemType data[maxsize] : 큐 데이터 배열

- Pop 연산
    - isEmpty() 가 false이면, front를 +1을 해줌
    - 만약, front == rear 인 경우, isEmpty()가 큐가 빈 것으로 처리되어서 실제 데이터를 삭제하지 않고도 데이터를 삭제한 것 처럼 관리 가능
- Push 연산
    - isFull()을 수행해서 큐가 가득 차지 않았다면 데이터 push
    - isFull()이 True 인 경우, 더 이상 push 하지 못함 

- Queue가 관리하는 공간은 front의 다음부터, rear까지의 데이터이고, 한방향으로 관리하기 때문에 메모리 공간을 낭비할 수 있음

### **3. 큐 구현하기**

1. 리스트 사용
- push 연산은 append() 메서드 사용, pop 연산은 pop() 메서드 사용
    - pop()의 인수로 0을 넣어야함 (선입선출!)


```python
queue = []

queue.append(1)
queue.append(2)
queue.append(3)

# 맨 앞 데이터 제거
first_item = queue.pop(0)
print(first_item)

queue.append(4)
queue.append(5)

first_item = queue.pop(0)
print(first_item)
```

2. DeQue를 큐 처럼 사용하기
- 양 끝에서 삽입이나 삭제를 할 수 있는 큐 구현

```python
from collections import deque

queue = deque()

# 데이터 추가
queue.append(1)
queue.append(2)
queue.append(3)

# 맨 앞 데이터 제거 
first_item = queue.popleft()
print(first_item)

queue.append(4)
queue.append(5)

first_item = queue.popleft()
print(first_item)
```

```python
# popleft() 와 pop(0) 시간 복잡도 비교하기

from collections import deque
import time

lst = list(range(100000))
dq = deque(range(100000))

# pop(0) 성능 측정
start_time = time.time()
for i in range(100000):
    lst.pop(0)

print("pop(0) 소요시간:", time.time() - start_time)

#popleft() 성능 측정
start_time = time.time()
for i in range(100000):
    dq.popleft()
print("popleft() 소요시간:", time.time() - start_time)
    
```
