## **문제 15: 요세푸스 문제**
- 시간복잡도 : O(N*K)
- 제약조건 : N, K는 1이상 1000이상의 자연수 
- 마지막 남은 요소 반환

```python
from collections import deque

def solution(N,K):
    queue = deque(range(1, N+1)) # O(N) 소요
    
    while  len(queue) > 1: # 마지막 1개의 요소가 남을 때 까지 반복 O(N-1)
        for i in range(K-1):  # for loop - O(K-1)
            queue.append(queue.popleft()) #popleft(), append()는 각각 O(1)
        
        queue.popleft()
        
    return queue[0] #마지막요소 반환
```

```python
print(solution(5,2))
```

#### 문제 풀이
- 마지막 1개의 원소가 남을 때 까지 계산을 해야하기 때문에 while loop 사용 - O(N-1) 
- deque를 사용해서 k번째에 있는 사람을 popleft() 위치로 옮기는 방식으로 해결가능
    - k-1번째 데이터까지 popleft() 후 다시 append() 를 해주면 k번째 데이터가 큐의 front에 위치하게 됨
    - for loop를 사용해서 구현하면 O(K-1)로 주어진 시간복잡도에 맞게 구현가능
